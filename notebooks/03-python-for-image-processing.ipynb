## 2. Names, objects and types

In Python, a name is bound to an object.

Unlike C, you do not normally declare a variable with a fixed type before assigning a value.


````python
width = 640

print(width)
print(type(width))
print(id(width))
````

`id()` demonstrates that Python objects have an identity. The returned value should not be treated as a portable memory address.

A name may later be rebound to an object of another type.


````python
value = 10
print(value, type(value))

value = 10.5
print(value, type(value))

value = "ten"
print(value, type(value))
````

### Historical C comparison

A C declaration fixes the declared object's type:

```c
int width = 640;
```

Python instead binds the name `width` to an integer object:

```python
width = 640
```

Python is dynamically typed, but Python objects themselves still have definite types.

## 5. Division in Python and C

Python's `/` operator performs true division:


````python
x = 3
y = 5

print(x / y)
print(x // y)
````

The output demonstrates the distinction:

- `/` performs true division;
- `//` performs floor division.

This differs from C integer division, where `3 / 5` with two integer operands produces `0`.

### C and Python block syntax

A C conditional commonly uses braces:

```c
if (pixel > 127)
{
    output = 255;
}
else
{
    output = 0;
}
```

Python uses indentation:

```python
if pixel > 127:
    output = 255
else:
    output = 0
```

Indentation is therefore part of Python's syntax.

## 19. The first reusable utility: `clamp_u8()`

Many image-processing operations produce intermediate values outside the valid unsigned 8-bit range:

\[
0 \leq p \leq 255
\]

A simple clamping function will be useful throughout the project.


````python
def clamp_u8(value):
    if value < 0:
        return 0

    if value > 255:
        return 255

    return value
````

````python
print(clamp_u8(-20))
print(clamp_u8(100))
print(clamp_u8(300))
````

Expected output:

```text
0
100
255
```


## 20. A reconstructed C equivalent

A newly written C implementation of the same idea is:

```c
unsigned char clamp_u8(int value)
{
    if (value < 0)
        return 0;

    if (value > 255)
        return 255;

    return (unsigned char) value;
}
```

This is a reconstruction for the book, not copied source code from the original UQC146S1 material.

The C version must explicitly specify the argument and return types and explicitly cast the final result back to `unsigned char`.


## 21. Type hints


````python
def clamp_u8(value: int) -> int:
    if value < 0:
        return 0

    if value > 255:
        return 255

    return value
````

Type hints document the intended types but do not normally enforce them at runtime.


## 22. Docstrings


````python
def clamp_u8(value: int) -> int:
    """Clamp an integer to the unsigned 8-bit range.

    Parameters
    ----------
    value:
        Integer value to clamp.

    Returns
    -------
    int
        0 if value is below 0, 255 if value is above
        255, otherwise value unchanged.
    """
    if value < 0:
        return 0

    if value > 255:
        return 255

    return value
````

## 24. Immutable objects

## 25. Mutable objects


````python
def set_first_pixel(pixels):
    pixels[0] = 255


values = [0, 64, 128]

set_first_pixel(values)

print(values)
````

The list is mutable, and the caller and function parameter can refer to the same list object.

This is one of the most important Python concepts to understand before passing NumPy arrays into image-processing functions.


## 26. Copies


````python
original = [0, 64, 128]
result = original.copy()

result[0] = 255

print(f"original = {original}")
print(f"result   = {result}")
````

The copied list can now be modified independently.


## 27. Basic validation and exceptions

The permanent utility should reject inappropriate input rather than silently produce an unexpected result.


````python
def clamp_u8(value: int) -> int:
    """Clamp an integer to the unsigned 8-bit range."""
    if not isinstance(value, int):
        raise TypeError("value must be an integer")

    if value < 0:
        return 0

    if value > 255:
        return 255

    return value
````

````python
print(clamp_u8(100))
````

The following call is deliberately invalid:


````python
try:
    clamp_u8("100")
except TypeError as error:
    print(f"TypeError: {error}")
````

Library code should generally raise a meaningful exception and allow the calling code to decide how to handle it.


## 28. Assertions as simple checks


````python
assert clamp_u8(-1) == 0
assert clamp_u8(0) == 0
assert clamp_u8(100) == 100
assert clamp_u8(255) == 255
assert clamp_u8(256) == 255

print("All clamp_u8 assertions passed.")
````

## 29. Bitwise operations

Binary operations will become important for packed image data and Huffman coding.


````python
value = 0b10110100
mask = 0b00000100

print(f"value = {value:08b}")
print(f"mask  = {mask:08b}")

if value & mask:
    print("The selected bit is set")
else:
    print("The selected bit is clear")
````

````python
a = 0b10101010
b = 0b11001100

print(f"AND: {a & b:08b}")
print(f"OR:  {a | b:08b}")
print(f"XOR: {a ^ b:08b}")
````

### Exercise A — classify a pixel

Write a function called `classify_pixel()` that returns:

- `"dark"` for values below 85;
- `"mid-tone"` for values from 85 to 169;
- `"bright"` for values from 170 upward.


````python
def classify_pixel(pixel: int) -> str:
    if pixel < 85:
        return "dark"

    if pixel < 170:
        return "mid-tone"

    return "bright"


print(classify_pixel(20))
print(classify_pixel(120))
print(classify_pixel(220))
````

### Exercise B — calculate image dimensions

Use the nested list image below to determine its height and width.


````python
image = [
    [0, 64, 128],
    [32, 96, 160],
    [64, 128, 255],
]

height = len(image)
width = len(image[0])

print(f"Width:  {width}")
print(f"Height: {height}")
````

### Exercise C — find the brightest pixel

Find the largest value in the nested image without using NumPy.


````python
brightest = image[0][0]

for row in image:
    for pixel in row:
        if pixel > brightest:
            brightest = pixel

print(f"Brightest pixel: {brightest}")
````

### Exercise D — invert a list of pixels

For an 8-bit greyscale pixel, a simple inversion is:

\[
p' = 255 - p
\]

Apply this to a one-dimensional list.


````python
pixels = [0, 64, 128, 192, 255]

inverted = []

for pixel in pixels:
    inverted.append(255 - pixel)

print(inverted)
````

### Exercise E — count pixels above a threshold


````python
threshold = 127
count = 0

for row in image:
    for pixel in row:
        if pixel > threshold:
            count += 1

print(f"Pixels above {threshold}: {count}")
````
